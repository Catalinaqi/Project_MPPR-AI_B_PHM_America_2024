```python
import numpy as np
import pandas as pd

# 1. Contar cuántos valores Infinitos o NaNs hay por columna
print("--- Conteo de Infinitos y Nulos por Columna ---")
for col in df_duck.columns:
    # Contar infinitos (positivos y negativos)
    inf_count = np.isinf(df_duck[col]).sum()
    # Contar Nulos (NaN)
    nan_count = df_duck[col].isna().sum()

    if inf_count > 0 or nan_count > 0:
        print(f"Columna '{col}': {inf_count} Infinitos, {nan_count} Nulos")

# 2. Ver un ejemplo de las filas que están rotas
print("\n--- Muestra de filas con problemas ---")
# Filtramos el dataframe para mostrar solo las filas que tengan al menos un infinito o un nulo
bad_rows = df_duck[df_duck.isin([np.inf, -np.inf]).any(axis=1) | df_duck.isna().any(axis=1)]

# Mostramos las primeras 5 filas defectuosas para que veas qué pasó
display(bad_rows.head())
```

```python
# Create a summary dataframe showing the Min and Max for every column
summary_df = pd.DataFrame({
    'Minimum': df_duck.min(),
    'Maximum': df_duck.max()
})

# Display the columns sorted by their maximum value in descending order
print("--- Check for Extreme Outliers (Potential Overflows) ---")
display(summary_df.sort_values(by='Maximum', ascending=False))
```
